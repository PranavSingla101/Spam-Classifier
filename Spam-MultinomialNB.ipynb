```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df = pd.read_csv('mail_data.csv')
```

```python
data = df.where((pd.notnull(df)), '')
```

```python
data.loc[data['Category'] == 'spam', 'Category',] = 0
data.loc[data['Category'] == 'ham', 'Category',] = 1
```

```python
X = data['Message']
Y = data['Category'].astype('int')
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=3)

print(f"Total data: {X.shape[0]}, Training data: {X_train.shape[0]}, Testing data: {X_test.shape[0]}")
```

```python
feature_extraction = TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)
```

```python
X_train_features = feature_extraction.fit_transform(X_train)
```

```python
X_test_features = feature_extraction.transform(X_test)
```

```python
print("Text data converted to TF-IDF features.")
```

```python
model = MultinomialNB()
```

```python
print("Training the MultinomialNB model...")
model.fit(X_train_features, Y_train)
print("Model trained.")
```

```python
prediction_on_training_data = model.predict(X_train_features)
accuracy_on_training_data = accuracy_score(Y_train, prediction_on_training_data)
```

```python
prediction_on_test_data = model.predict(X_test_features)
accuracy_on_test_data = accuracy_score(Y_test, prediction_on_test_data)
```

```python
print("\n--- Model Evaluation ---")
print(f"Accuracy on training data: {accuracy_on_training_data * 100:.2f}%")
print(f"Accuracy on test data: {accuracy_on_test_data * 100:.2f}%")
```

```python
print("\nClassification Report (Test Data):")
# target_names=['Spam (0)', 'Ham (1)'] matches our 0/1 encoding
print(classification_report(Y_test, prediction_on_test_data, target_names=['Spam (0)', 'Ham (1)']))
```

```python
print("\nConfusion Matrix (Test Data):")
print(confusion_matrix(Y_test, prediction_on_test_data))
```

```python
input_data_features = feature_extraction.transform(input_your_mail)
```

```python
prediction = model.predict(input_data_features)
```
